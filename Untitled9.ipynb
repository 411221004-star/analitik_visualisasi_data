```python
import sqlite3
import pandas as pd

# Membuat koneksi ke database SQLite
conn = sqlite3.connect('perpustakaan.db')
cursor = conn.cursor()

# Mengaktifkan dukungan Foreign Key di SQLite
cursor.execute("PRAGMA foreign_keys = ON;")

# 1. Membuat tabel buku
cursor.execute('''
CREATE TABLE IF NOT EXISTS buku (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    judul TEXT NOT NULL,
    penulis TEXT NOT NULL,
    tahun INTEGER
);
''')

# 2. Membuat tabel anggota
cursor.execute('''
CREATE TABLE IF NOT EXISTS anggota (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nama TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL
);
''')

# 3. Membuat tabel peminjaman
cursor.execute('''
CREATE TABLE IF NOT EXISTS peminjaman (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    id_anggota INTEGER,
    id_buku INTEGER,
    tgl_pinjam DATE NOT NULL,
    tgl_kembali DATE,
    FOREIGN KEY (id_anggota) REFERENCES anggota(id),
    FOREIGN KEY (id_buku) REFERENCES buku(id)
);
''')

conn.commit()
print("Tabel berhasil dibuat!")
```

```python
# Menutup koneksi lama secara paksa untuk melepas kunci (lock)
try:
    conn.close()
except:
    pass

# Membuka koneksi baru
import sqlite3
conn = sqlite3.connect('perpustakaan.db')
cursor = conn.cursor()
cursor.execute("PRAGMA foreign_keys = ON;")

# 1a. Menambahkan 5 data buku (menggunakan OR IGNORE agar aman jika dijalankan ulang)
data_buku = [
    ('Laskar Pelangi', 'Andrea Hirata', 2005),
    ('Bumi Manusia', 'Pramoedya Ananta Toer', 1980),
    ('Filosofi Teras', 'Henry Manampiring', 2018),
    ('Clean Code', 'Robert C. Martin', 2008),
    ('Atomic Habits', 'James Clear', 2018)
]
cursor.executemany('INSERT OR IGNORE INTO buku (judul, penulis, tahun) VALUES (?, ?, ?)', data_buku)

# 1a. Menambahkan 3 data anggota (menggunakan OR IGNORE untuk menghindari UNIQUE constraint error pada email)
data_anggota = [
    ('Budi Santoso', 'budi.santoso@email.com'),
    ('Siti Aminah', 'siti.aminah@email.com'),
    ('Andi Wijaya', 'andi.wijaya@email.com')
]
cursor.executemany('INSERT OR IGNORE INTO anggota (nama, email) VALUES (?, ?)', data_anggota)

# 1b. Menambahkan 4 transaksi peminjaman (menggunakan OR IGNORE agar tidak duplikat)
data_peminjaman = [
    (1, 1, '2026-11-01', '2026-11-05'),  # Budi meminjam Laskar Pelangi (4 hari)
    (1, 3, '2026-11-10', '2026-11-17'),  # Budi meminjam Filosofi Teras (7 hari)
    (2, 2, '2026-11-12', '2026-11-15'),  # Siti meminjam Bumi Manusia (3 hari)
    (3, 4, '2026-11-18', None)           # Andi meminjam Clean Code (Belum dikembalikan / NULL)
]
cursor.executemany('INSERT OR IGNORE INTO peminjaman (id_anggota, id_buku, tgl_pinjam, tgl_kembali) VALUES (?, ?, ?, ?)', data_peminjaman)

conn.commit()
print("Data buku, anggota, dan peminjaman berhasil diproses tanpa duplikasi!")
```

```python
# 1c. Menampilkan daftar peminjaman lengkap menggunakan JOIN
query_1c = '''
SELECT
    p.id AS id_peminjaman,
    a.nama AS nama_anggota,
    b.judul AS judul_buku,
    p.tgl_pinjam,
    p.tgl_kembali
FROM peminjaman p
JOIN anggota a ON p.id_anggota = a.id
JOIN buku b ON p.id_buku = b.id;
'''
print("=== 1c. Daftar Peminjaman Lengkap ===")
df_peminjaman = pd.read_sql_query(query_1c, conn)
print(df_peminjaman.to_string(index=False))

print("\n" + "="*40 + "\n")

# 1d. Menghitung jumlah total buku yang dipinjam oleh masing-masing anggota
query_1d = '''
SELECT
    a.nama AS nama_anggota,
    COUNT(p.id) AS total_buku_dipinjam
FROM anggota a
LEFT JOIN peminjaman p ON a.id = p.id_anggota
GROUP BY a.id, a.nama
ORDER BY total_buku_dipinjam DESC;
'''
print("=== 1d. Total Buku Dipinjam per Anggota ===")
df_total_pinjam = pd.read_sql_query(query_1d, conn)
print(df_total_pinjam.to_string(index=False))
```

```python
# 2a. DataFrame sudah dibuat dari query SQL pada langkah 1c (df_peminjaman)
df_eksplorasi = df_peminjaman.copy()

# 2b. Eksplorasi data: Mengonversi kolom tanggal ke datetime dan menghitung durasi (hari)
df_eksplorasi['tgl_pinjam'] = pd.to_datetime(df_eksplorasi['tgl_pinjam'])
df_eksplorasi['tgl_kembali'] = pd.to_datetime(df_eksplorasi['tgl_kembali'])

# Menghitung selisih hari (hasilnya akan NaN jika tgl_kembali masih NULL)
df_eksplorasi['durasi_hari'] = (df_eksplorasi['tgl_kembali'] - df_eksplorasi['tgl_pinjam']).dt.days

# Menambahkan kolom status untuk mempermudah identifikasi pola
df_eksplorasi['status'] = df_eksplorasi['tgl_kembali'].apply(
    lambda x: 'Belum Dikembalikan' if pd.isna(x) else 'Sudah Dikembalikan'
)

print("=== Hasil Eksplorasi Data Peminjaman ===")
print(df_eksplorasi[['nama_anggota', 'judul_buku', 'durasi_hari', 'status']].to_string(index=False))

# Statistik ringkas
rata_rata_durasi = df_eksplorasi['durasi_hari'].mean()
belum_kembali = df_eksplorasi['tgl_kembali'].isna().sum()

print(f"\nRata-rata durasi peminjaman (yang sudah kembali): {rata_rata_durasi:.2f} hari")
print(f"Jumlah buku yang masih dipinjam (belum kembali): {belum_kembali} buku")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur gaya visualisasi
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Grafik 1: Diagram Batang Jumlah Buku yang Dipinjam per Anggota
sns.barplot(
    data=df_total_pinjam,
    x='nama_anggota',
    y='total_buku_dipinjam',
    hue='nama_anggota',      # Ditambahkan untuk menyesuaikan aturan Seaborn terbaru
    palette='viridis',
    legend=False,            # Ditambahkan agar legenda tidak muncul ganda
    ax=axes[0]
)
axes[0].set_title('Jumlah Buku Dipinjam per Anggota', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Nama Anggota', fontsize=10)
axes[0].set_ylabel('Total Buku', fontsize=10)
axes[0].set_yticks(range(0, df_total_pinjam['total_buku_dipinjam'].max() + 2))

# Menambahkan label angka di atas batang Grafik 1
for p in axes[0].patches:
    axes[0].annotate(f'{int(p.get_height())}',
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='bottom', fontsize=10, fontweight='bold')

# Grafik 2: Durasi Peminjaman Buku (Hanya untuk buku yang sudah dikembalikan)
df_kembali = df_eksplorasi.dropna(subset=['durasi_hari'])
sns.barplot(
    data=df_kembali,
    x='judul_buku',
    y='durasi_hari',
    hue='judul_buku',        # Ditambahkan untuk menyesuaikan aturan Seaborn terbaru
    palette='magma',
    legend=False,            # Ditambahkan agar legenda tidak muncul ganda
    ax=axes[1]
)
axes[1].set_title('Durasi Peminjaman per Buku (Hari)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Judul Buku', fontsize=10)
axes[1].set_ylabel('Durasi (Hari)', fontsize=10)
axes[1].set_yticks(range(0, int(df_kembali['durasi_hari'].max()) + 3))

# Menambahkan label angka di atas batang Grafik 2
for p in axes[1].patches:
    axes[1].annotate(f'{int(p.get_height())} hari',
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```
